```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# -------------------------------------------------
# 1. Load IMDB dataset
# -------------------------------------------------

vocab_size = 10000
max_len = 200

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)

X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)

# -------------------------------------------------
# 2. Build simple LSTM model
# -------------------------------------------------

model = Sequential([
    Embedding(vocab_size, 128, input_length=max_len),
    LSTM(64),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# -------------------------------------------------
# 3. Train model
# -------------------------------------------------

model.fit(X_train, y_train, epochs=2, batch_size=64)

# -------------------------------------------------
# 4. Load word dictionary
# -------------------------------------------------

word_index = imdb.get_word_index()

# -------------------------------------------------
# 5. Function to encode sentence
# -------------------------------------------------

def encode_review(text):
    
    # lowercase
    text = text.lower()

    # split sentence into words
    words = text.split()

    # convert words to numbers
    encoded = []

    for word in words:
        if word in word_index:
            encoded.append(word_index[word] + 3)
        else:
            encoded.append(2)   # unknown word

    return encoded

# -------------------------------------------------
# 6. Input your own sentence
# -------------------------------------------------

my_review = "this movie was amazing and wonderful"

encoded_review = encode_review(my_review)

# padding
padded_review = pad_sequences(
    [encoded_review],
    maxlen=max_len
)

# -------------------------------------------------
# 7. Predict
# -------------------------------------------------

prediction = model.predict(padded_review)

print("Prediction:", prediction[0][0])

if prediction[0][0] > 0.5:
    print("Positive Review")
else:
    print("Negative Review")
```
